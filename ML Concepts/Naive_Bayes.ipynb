```python
if 'df' not in locals() or not isinstance(df, pd.DataFrame):
    df = pd.read_csv("/content/WA_Fn-UseC_-Telco-Customer-Churn.csv")
df = df.drop(['Partner', 'Dependents', 'PhoneService',
       'MultipleLines', 'InternetService', 'OnlineSecurity', 'OnlineBackup',
       'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies', 'PaperlessBilling',  'TotalCharges'],axis=1, errors='ignore')
```

```python
df.head()
```

```python
X = df.drop(["customerID", "Churn"], axis=1)
y = df["Churn"]
```

```python
for col in df.columns:
    print("\n", col)
    print(df[col].value_counts())
```

```python
df["Churn"].value_counts()
```

```python
df["Churn"].value_counts(normalize=True) * 100
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training data:", X_train.shape)
print("Testing data :", X_test.shape)
```

```python
categorical_cols = ['gender', 'Contract', 'PaymentMethod']

numerical_cols = ['SeniorCitizen', 'tenure', 'MonthlyCharges']
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
        ("num", StandardScaler(), numerical_cols)
    ]
)
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.pipeline import Pipeline

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", GaussianNB())
])
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
import matplotlib.pyplot as plt

ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
plt.show()
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
y_prob = model.predict_proba(X_test)[:, 1]
from sklearn.metrics import roc_auc_score

auc = roc_auc_score(y_test, y_prob)

print("ROC-AUC:", auc)
```

```python
print(classification_report(y_test, y_pred))

print("ROC-AUC:", roc_auc_score(y_test, y_prob))
```
