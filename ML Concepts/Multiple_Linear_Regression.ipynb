```python
X = df.drop('Price_lakh',axis = 1)
y = df['Price_lakh']
```

```python
from sklearn.model_selection import train_test_split
X_train , X_test , y_train , y_test = train_test_split(X,y,random_state=42,test_size=0.2)
```

```python
y_test.shape
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train,y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
from sklearn.metrics import r2_score , mean_absolute_error ,mean_squared_error,root_mean_squared_error
print(r2_score(y_test,y_pred))

print(mean_absolute_error(y_test,y_pred))

print(mean_squared_error(y_test,y_pred))

print(root_mean_squared_error(y_test,y_pred))
```

```python
for features , coefficient in zip(X_train.columns,lr.coef_):
  print(features,coefficient)
```

```python
from sklearn.metrics import r2_score

# 1. Get predictions for both datasets
y_train_pred = lr.predict(X_train)
y_test_pred = lr.predict(X_test)

# 2. Calculate R2 score for training data
train_r2 = r2_score(y_train, y_train_pred)

# 3. Calculate R2 score for testing data
test_r2 = r2_score(y_test, y_test_pred)

print(f"Train R2 Score: {train_r2}")
print(f"Test R2 Score: {test_r2}")
```
